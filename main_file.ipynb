```python
import geopandas as gpd 

bg = gpd.read_file("data/raw/tl_2025_48_bg.zip")

print(bg.shape)
print(bg.crs)
print(bg.columns.tolist())
print(bg.head())
```

```python
print(bg['COUNTYFP'].dtype)
```

```python
bg_Harris =bg[bg['COUNTYFP'] =='201']
bg_Harris.head()
bg_Harris.shape
```

```python
bg_Harris.to_file("data/processed/harris_block_groups.gpkg",driver="GPKG")
```

```python
harris_boundary = bg_Harris.dissolve()
print(harris_boundary.shape)
```

```python
import osmnx as ox

harris_geom = harris_boundary.geometry.iloc[0]

substations = ox.features_from_polygon(harris_geom,tags={"power":"substation"})
print(substations.shape)
print(substations.crs)
print(substations.geometry.geom_type.value_counts())
```
